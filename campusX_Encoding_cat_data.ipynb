```python
import pandas as pd
import numpy as np
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv('covid_toy.csv')
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test= train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

```python
X_train
```

## **Encoding categorical data**





categorical to numerical columns cahenges differently

```python
# Adding Simple Imputer
si=SimpleImputer()
X_train_fever=si.fit_transform(X_train[['fever']])
X_test_fever=si.fit_transform(X_test[['fever']])
```

```python
X_train_fever.shape
```

```python
# Ordinal encoder
oe=OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough=oe.fit_transform(X_train[['cough']])

X_test_cough=oe.fit_transform(X_test[['cough']])
```

```python
X_train_cough.shape
```

```python
# One hot Encoding

ohe=OneHotEncoder(drop='first',sparse_output=False)

X_train_gender_city=ohe.fit_transform(X_train[['gender','city']])
X_test_gender_city=ohe.fit_transform(X_test[['gender','city']])
```

```python
X_train_gender_city.shape
```

```python
# Extracting Age

X_train_age=X_train.drop(columns=['gender','city','cough','fever']).values
X_test_age=X_test.drop(columns=['gender','city','cough','fever']).values
```

```python
X_train_age.shape
```

```python
# Concatenate all tranformed columns

X_train_transformed=np.concatenate((X_train_age,X_train_gender_city,X_train_fever,X_train_cough),axis=1)

X_test_transformed=np.concatenate((X_test_age,X_test_gender_city,X_test_fever,X_test_cough),axis=1)
```

# **Column Transformer**

all transformation happens in single line

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer=ColumnTransformer(transformers=[('tnf1',SimpleImputer(),['fever']),
                                            ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
                                            ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
                                            ],remainder='passthrough')
```

```python
X_trans_train=transformer.fit_transform(X_train)
X_trans_test=transformer.fit_transform(X_test)
```
