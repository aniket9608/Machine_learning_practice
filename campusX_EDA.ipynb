```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('train.csv')
```

```python
df.head()
```

```python
# countplot

sns.countplot(df['Embarked'])
# df['Survived'].value_counts().plot(kind='bar')
```

```python
# piechart
df['Sex'].value_counts().plot(kind='pie',autopct='%2f')
```

```python
# Categorical Data
# Histogram
import matplotlib.pyplot as plt

plt.hist(df['Age'],bins=20)
```

```python
# Displot

sns.displot(df['Age'])
```

```python
# Boxplot

sns.boxplot(df['Age'])
```

**BIVARIATE**

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
# Scatterplot (n-n)

sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=df['Sex'])
```

```python
# barplot (n-c)

sns.barplot(x=df['Pclass'],y=df['Age'],hue=df['Sex'])
```

```python
# Boxplot(n-c)

sns.boxplot(x=df['Pclass'],y=df['Age'],hue=df['Survived'])
```

```python
# Displot (n-c)

sns.displot(x=df[df['Survived']==0]['Age'],kind='kde')
sns.displot(x=df[df['Survived']==1]['Age'],kind='kde')
```

```python
# HeatMap

sns.heatmap(pd.crosstab(df['Pclass'],df['Survived']))
```

```python
# ClusterMap

sns.clustermap(pd.crosstab(df['Pclass'],df['Survived']))
```
